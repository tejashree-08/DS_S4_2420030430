```python
# Week 9.1
# Probability Distributions
# 1. Bernoulli Distribution
# 2. Binomial Distribution
# Application: Student Exam Pass/Fail Analysis

import numpy as np
from scipy.stats import bernoulli, binom

# ---------------------------------------------------
# BERNOULLI DISTRIBUTION
# ---------------------------------------------------

print("----- Bernoulli Distribution -----")

# Probability of passing the exam
# 0.7 means 70% chance of passing
p = 0.7

# Generate 10 random Bernoulli outcomes
# 1 = Pass
# 0 = Fail
bernoulli_samples = bernoulli.rvs(p, size=10)

print("Probability of Passing:", p)

print("Student Outcomes (1=Pass, 0=Fail):")
print(bernoulli_samples)

# Calculate probability of success/pass
# P(X = 1) = p
print("\nProbability of Pass:")
print(bernoulli.pmf(1, p))

# Calculate probability of failure
# P(X = 0) = 1 - p
print("Probability of Fail:")
print(bernoulli.pmf(0, p))


# ---------------------------------------------------
# BINOMIAL DISTRIBUTION
# ---------------------------------------------------

print("\n\n----- Binomial Distribution -----")

# Number of students/trials
n = 10

# Probability of passing each exam
p = 0.7

# Number of students we want to find probability for
k = 8

# Calculate probability that exactly 8 students pass
binomial_probability = binom.pmf(k, n, p)

print("Number of Students:", n)
print("Probability of Passing:", p)

print(f"\nProbability that exactly {k} students pass:")
print(binomial_probability)

# Generate 5 random binomial experiments
# Each experiment represents 10 students
random_binomial = binom.rvs(n, p, size=5)

print("\nRandom Binomial Outcomes:")
print(random_binomial)
```

```python
# Week 9.2
# Statistical Hypothesis Testing
# 1. Z-Test
# 2. Independent t-Test
# Application: Student Performance Analysis

import numpy as np
from scipy import stats
from statsmodels.stats.weightstats import ztest


# ---------------------------------------------------
# Z-TEST
# ---------------------------------------------------

print("----- Z-Test -----")

# Marks obtained by 10 students
sample_marks = np.array([
    78, 82, 85, 88, 90,
    76, 95, 89, 84, 91
])

# Population average mark
population_mean = 80

# Perform one-sample Z-test
# It checks whether sample mean is significantly
# different from population mean
z_stat, p_value = ztest(
    sample_marks,
    value=population_mean
)

print("Sample Marks:")
print(sample_marks)

print("\nPopulation Mean:", population_mean)

# Display Z statistic
print("\nZ-Statistic:", z_stat)

# Display p-value
print("P-Value:", p_value)

# Significance level
alpha = 0.05

# Decision:
# If p-value < 0.05, reject null hypothesis
if p_value < alpha:
    print("\nResult: Reject Null Hypothesis")
else:
    print("\nResult: Fail to Reject Null Hypothesis")


# ---------------------------------------------------
# INDEPENDENT t-TEST
# ---------------------------------------------------

print("\n\n----- Independent t-Test -----")

# Marks of Class A
class_A = np.array([
    78, 82, 85, 88, 90
])

# Marks of Class B
class_B = np.array([
    72, 75, 70, 68, 74
])

# Perform independent t-test
# It compares the means of two independent groups
t_stat, p_value = stats.ttest_ind(
    class_A,
    class_B
)

print("Class A Marks:")
print(class_A)

print("\nClass B Marks:")
print(class_B)

# Display t statistic
print("\nt-Statistic:", t_stat)

# Display p-value
print("P-Value:", p_value)

# Make decision using significance level
if p_value < alpha:
    print("\nResult: Reject Null Hypothesis")
else:
    print("\nResult: Fail to Reject Null Hypothesis")
```
