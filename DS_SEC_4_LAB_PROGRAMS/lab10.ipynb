```python
# Week 10
# Regression Models for Predictive Analytics
# 1. Linear Regression
# 2. Logistic Regression
#
# Application:
# House Price Prediction
# House Purchase Classification

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression


# ---------------------------------------------------
# LINEAR REGRESSION
# ---------------------------------------------------

print("----- Linear Regression -----")

# Independent variable
# House size in square feet
X_linear = np.array([
    [1000],
    [1200],
    [1500],
    [1800],
    [2000]
])

# Dependent variable
# House price in lakhs
y_linear = np.array([
    50, 55, 65, 80, 90
])

# Create Linear Regression model
linear_model = LinearRegression()

# Train the model using house size and price
linear_model.fit(
    X_linear,
    y_linear
)

# New house size for prediction
new_house_size = np.array([
    [1700]
])

# Predict price of 1700 sq.ft house
predicted_price = linear_model.predict(
    new_house_size
)

print("Training House Sizes:")
print(X_linear.flatten())

print("\nTraining House Prices:")
print(y_linear)

print("\nPredicted Price for 1700 sq.ft house:")
print(predicted_price[0])


# ---------------------------------------------------
# LOGISTIC REGRESSION
# ---------------------------------------------------

print("\n\n----- Logistic Regression -----")

# House prices used for training
X_logistic = np.array([
    [40],
    [50],
    [60],
    [70],
    [80],
    [90]
])

# Output labels
# 0 = Not Purchased
# 1 = Purchased
y_logistic = np.array([
    0, 0, 0, 1, 1, 1
])

# Create Logistic Regression model
logistic_model = LogisticRegression()

# Train the model
logistic_model.fit(
    X_logistic,
    y_logistic
)

# Test house price
test_price = np.array([
    [75]
])

# Predict whether customer will purchase
purchase_prediction = logistic_model.predict(
    test_price
)

# Find probability of each class
purchase_probability = logistic_model.predict_proba(
    test_price
)

print("Training House Prices:")
print(X_logistic.flatten())

print("\nPurchase Labels:")
print(y_logistic)

print("\nTest House Price:")
print(test_price[0][0])

print("\nPurchase Prediction:")
print(purchase_prediction[0])

print("\nPrediction Probabilities:")
print(purchase_probability)
```
