```python
# Week 11.1
# Classification using Decision Tree
# Performance Evaluation Metrics
# Application: Student Pass/Fail Prediction

import numpy as np

from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)


# ---------------------------------------------------
# INPUT DATA
# ---------------------------------------------------

# Features:
# First column  = Study Hours
# Second column = Attendance Percentage

X = np.array([
    [2, 50],
    [3, 60],
    [4, 65],
    [5, 70],
    [6, 75],
    [7, 80],
    [8, 85],
    [9, 90]
])

# Output labels:
# 0 = Fail
# 1 = Pass

y = np.array([
    0, 0, 0, 1,
    1, 1, 1, 1
])


# ---------------------------------------------------
# CREATE DECISION TREE MODEL
# ---------------------------------------------------

# Create Decision Tree classifier
model = DecisionTreeClassifier(
    random_state=42
)

# Train the model
model.fit(X, y)


# ---------------------------------------------------
# TEST DATA
# ---------------------------------------------------

# Test student data
# [Study Hours, Attendance]

X_test = np.array([
    [3, 55],
    [5, 72],
    [8, 88],
    [4, 60]
])

# Actual test results
# 0 = Fail
# 1 = Pass

y_test = np.array([
    0, 1, 1, 0
])


# ---------------------------------------------------
# MAKE PREDICTIONS
# ---------------------------------------------------

# Predict Pass/Fail for test students
y_pred = model.predict(X_test)


# ---------------------------------------------------
# PERFORMANCE EVALUATION
# ---------------------------------------------------

# Accuracy:
# Percentage of correct predictions
accuracy = accuracy_score(
    y_test,
    y_pred
)

# Error rate:
# Percentage of incorrect predictions
error_rate = 1 - accuracy

# Precision:
# Correct positive predictions
precision = precision_score(
    y_test,
    y_pred
)

# Recall:
# Correctly identified actual positive cases
recall = recall_score(
    y_test,
    y_pred
)

# F1-score:
# Combination of precision and recall
f1 = f1_score(
    y_test,
    y_pred
)

# Confusion matrix:
# Shows correct and incorrect classifications
cm = confusion_matrix(
    y_test,
    y_pred
)


# ---------------------------------------------------
# DISPLAY RESULTS
# ---------------------------------------------------

print("Actual Labels:")
print(y_test)

print("\nPredicted Labels:")
print(y_pred)

print("\nConfusion Matrix:")
print(cm)

print("\nAccuracy:")
print(accuracy)

print("\nError Rate:")
print(error_rate)

print("\nPrecision:")
print(precision)

print("\nRecall:")
print(recall)

print("\nF1-Score:")
print(f1)
```

```python
# Week 11.2
# K-Nearest Neighbors (KNN) Classification
# Performance Evaluation Metrics
# Application: Student Pass/Fail Prediction

import numpy as np

from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)


# ---------------------------------------------------
# TRAINING DATA
# ---------------------------------------------------

# Features:
# Study Hours
# Attendance Percentage

X_train = np.array([
    [2, 50],
    [3, 60],
    [4, 65],
    [5, 70],
    [6, 75],
    [7, 80],
    [8, 85],
    [9, 90]
])

# Labels:
# 0 = Fail
# 1 = Pass

y_train = np.array([
    0, 0, 0, 1,
    1, 1, 1, 1
])


# ---------------------------------------------------
# CREATE KNN MODEL
# ---------------------------------------------------

# n_neighbors=3 means
# the model looks at 3 nearest students
model = KNeighborsClassifier(
    n_neighbors=3
)

# Train KNN model
model.fit(
    X_train,
    y_train
)


# ---------------------------------------------------
# TEST DATA
# ---------------------------------------------------

# New student data
X_test = np.array([
    [3, 55],
    [5, 72],
    [8, 88],
    [4, 60]
])

# Actual results
y_test = np.array([
    0, 1, 1, 0
])


# ---------------------------------------------------
# PREDICTION
# ---------------------------------------------------

# Predict Pass/Fail
y_pred = model.predict(
    X_test
)


# ---------------------------------------------------
# PERFORMANCE EVALUATION
# ---------------------------------------------------

# Calculate accuracy
accuracy = accuracy_score(
    y_test,
    y_pred
)

# Calculate error rate
error_rate = 1 - accuracy

# Calculate precision
precision = precision_score(
    y_test,
    y_pred
)

# Calculate recall
recall = recall_score(
    y_test,
    y_pred
)

# Calculate F1-score
f1 = f1_score(
    y_test,
    y_pred
)

# Create confusion matrix
cm = confusion_matrix(
    y_test,
    y_pred
)


# ---------------------------------------------------
# DISPLAY RESULTS
# ---------------------------------------------------

print("Actual Labels:")
print(y_test)

print("\nPredicted Labels:")
print(y_pred)

print("\nConfusion Matrix:")
print(cm)

print("\nAccuracy:")
print(accuracy)

print("\nError Rate:")
print(error_rate)

print("\nPrecision:")
print(precision)

print("\nRecall:")
print(recall)

print("\nF1-Score:")
print(f1)
```
